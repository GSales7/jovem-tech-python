```python
usuario = "Pedro"

if usuario.lower() == "mariana":
  print("Bem-vinda, Mariana!")
elif usuario.lower() == "pedro":
  print("Bem-vindo, Pedro!")
else:
  print("Usuário não reconhecido!")
```
